```python
from pydantic import BaseModel, Field
from typing import List, Annotated
```

```python
class user(BaseModel):
    name : str
    age : int
```

```python
user=user(name="Krishna",age=27)
user
```

### Field Validation

```python
class Product(BaseModel):
    title:str = Field(max_length=10)
    des:str = Field(min_length=20)
    price:int =Field(gt=0,le=100)
```

```python
product =Product(title="SamsungS23", des = "This is amazing mobile from samsung...",price=20)
product2 =Product(title="Mobile", des = "This is amazing mobile from samsung...",price=20)
```

```python
class Order(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    product:Product
```

```python
order = Order(id="abc123",product=product)
```

```python
order
```

```python
order.product.title
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items:List[Product] #[product1, product2]
    comment: str = Field(
        default="No comment"
    )
```

```python
cart=Carts(id="ab1234", items=[product,product2],comment="This is my cart")
```

```python
cart
```

```python
for product in cart.items:
    print(product.title)
```

### Annotated

```python
commonValidator = Annotated[List[Product],Field(default=[])]
```

```python
class Carts(BaseModel):
    id:str=Field(min_length=6,max_length=12)
    items:commonValidator
    comment: str = Field(
        default="No comment"
    )
```

```python
class Order(BaseModel):
    products:commonValidator
```

```python
ord = Order(products=[product,product2])
```
