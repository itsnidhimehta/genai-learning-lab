### Load Env Variables

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_openai import ChatOpenAI
llm= ChatOpenAI(model="gpt-4")
```

```python
res=llm.invoke("Who was the winner of IPL 2020?")
print(res.content)
```

### Connect with Google Gemini Model
