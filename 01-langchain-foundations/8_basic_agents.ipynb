```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int,b:int):
    """
        it will return the sum of two numbers

        Args:
            a: Number one
            b: Number two
    
    """
    return a+b

@tool
def multiply_numbers(a:int,b:int):
    """
        it will return the product of two numbers

        Args:
            a: Number one
            b: Number two
    
    """
    return a*b
```

```python
add_numbers.invoke({"a":12,"b":3})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm=ChatOpenAI(model="gpt-4")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers,multiply_numbers],
    system_prompt="You are a math teacher, and always use tools for calculation"
)
```

```python
response=agent.invoke({"messages":[{"role":"user","content":"what is 2+30*10?"}]})
print(response["messages"][-1].content)
```
